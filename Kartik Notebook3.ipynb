```python
import pandas as pd
from IPython.display import display, Markdown, Math

pd.set_option("display.max_colwidth", 120)

def show_table(df, title):
    display(Markdown(f"## {title}"))
    display(
        df.style
        .hide(axis="index")
        .set_table_styles([
            {"selector": "th", "props": [
                ("background-color", "#f2f2f2"),
                ("font-weight", "bold"),
                ("text-align", "left"),
                ("border", "1px solid #cccccc"),
                ("padding", "6px")
            ]},
            {"selector": "td", "props": [
                ("text-align", "left"),
                ("border", "1px solid #dddddd"),
                ("padding", "6px")
            ]},
            {"selector": "table", "props": [
                ("border-collapse", "collapse"),
                ("width", "100%")
            ]}
        ])
    )

display(Markdown("# PFTR plot checklist"))

display(Markdown("## Dimensional variables"))
display(Math(r"z,\quad c_{EO},\quad c_{EG},\quad T,\quad X_{EO},\quad r"))

display(Markdown("## Heat terms"))
display(Math(r"q_{gen}= -\Delta_R H \cdot r"))
display(Math(r"q_{rem}= \frac{4h_{wall}}{d_R}\left(T-T_{wall}\right)"))

dimensional_base_plots = pd.DataFrame([
    {
        "Plot": "Concentration profiles",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$c_{EO},\ c_{EG}\ /\ \mathrm{mol\,m^{-3}}$"
    },
    {
        "Plot": "Temperature profile",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$T\ /\ \mathrm{K}$"
    },
    {
        "Plot": "Conversion profile",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$X_{EO}\ /\ -$"
    },
    {
        "Plot": "Reaction-rate profile",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$r\ /\ \mathrm{mol\,m^{-3}\,s^{-1}}$"
    },
    {
        "Plot": "Heat generation and heat removal",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$q_{gen},\ q_{rem}\ /\ \mathrm{W\,m^{-3}}$"
    },
    {
        "Plot": "Reaction path",
        "x-axis": r"$T\ /\ \mathrm{K}$",
        "y-axis": r"$X_{EO}\ /\ -$"
    }
])

show_table(dimensional_base_plots, "Dimensional base-case plots")

verification_plots = pd.DataFrame([
    {
        "Plot": "Mass-balance check",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$c_{EO}+c_{EG}\ /\ \mathrm{mol\,m^{-3}}$"
    },
    {
        "Plot": "Isothermal analytical validation",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$c_{EO}\ /\ \mathrm{mol\,m^{-3}}$"
    },
    {
        "Plot": "Adiabatic temperature check",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$T\ /\ \mathrm{K}$"
    },
    {
        "Plot": "Solver tolerance check",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$X_{EO}\ /\ -$"
    }
])

show_table(verification_plots, "Verification plots")

display(Markdown("## Validation equations"))
display(Math(r"c_{EO}(z)=c_{EO,in}\exp\left(-\frac{kz}{u}\right)"))
display(Math(r"X_{EO}(z)=\frac{c_{EO,in}-c_{EO}(z)}{c_{EO,in}}"))

dimensional_parameter_plots = pd.DataFrame([
    {
        "Plot": r"$T(z)$ for varying $T_{wall}$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$T\ /\ \mathrm{K}$"
    },
    {
        "Plot": r"$X_{EO}(z)$ for varying $T_{wall}$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$X_{EO}\ /\ -$"
    },
    {
        "Plot": r"$T(z)$ for varying $h_{wall}$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$T\ /\ \mathrm{K}$"
    },
    {
        "Plot": r"$X_{EO}(z)$ for varying $h_{wall}$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$X_{EO}\ /\ -$"
    },
    {
        "Plot": r"$T(z)$ for varying $d_R$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$T\ /\ \mathrm{K}$"
    },
    {
        "Plot": r"$X_{EO}(z)$ for varying $d_R$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$X_{EO}\ /\ -$"
    },
    {
        "Plot": r"$T(z)$ for varying $u$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$T\ /\ \mathrm{K}$"
    },
    {
        "Plot": r"$X_{EO}(z)$ for varying $u$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$X_{EO}\ /\ -$"
    },
    {
        "Plot": r"$T(z)$ for varying $T_{in}$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$T\ /\ \mathrm{K}$"
    },
    {
        "Plot": r"$X_{EO}(z)$ for varying $T_{in}$",
        "x-axis": r"$z\ /\ \mathrm{m}$",
        "y-axis": r"$X_{EO}\ /\ -$"
    }
])

show_table(dimensional_parameter_plots, "Dimensional parameter-study plots")

display(Markdown("## Dimensionless variables"))
display(Math(r"\zeta=\frac{z}{L}"))
display(Math(r"f_{EO}=\frac{c_{EO}}{c_{EO,in}}"))
display(Math(r"f_{EG}=\frac{c_{EG}}{c_{EO,in}}"))
display(Math(r"\vartheta=\frac{T}{T_{in}}"))
display(Math(r"X_{EO}=1-f_{EO}"))

display(Markdown("## Dimensionless groups"))
display(Math(r"Da_I=k(T_{in})\frac{L}{u}"))
display(Math(r"St=\frac{4h_{wall}L}{u d_R\rho c_p}"))
display(Math(r"\beta=\frac{-\Delta_RH\ c_{EO,in}}{\rho c_p T_{in}}"))
display(Math(r"\gamma=\frac{E_A}{RT_{in}}"))

dimensionless_base_plots = pd.DataFrame([
    {
        "Plot": "Dimensionless concentration profiles",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$f_{EO},\ f_{EG}\ /\ -$"
    },
    {
        "Plot": "Dimensionless temperature profile",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$\vartheta\ /\ -$"
    },
    {
        "Plot": "Dimensionless conversion profile",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$X_{EO}\ /\ -$"
    },
    {
        "Plot": "Dimensionless reaction-rate profile",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$\omega=r/r_{in}\ /\ -$"
    },
    {
        "Plot": "Dimensionless heat terms",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$\mathrm{dimensionless\ heat\ terms}\ /\ -$"
    },
    {
        "Plot": "Dimensionless reaction path",
        "x-axis": r"$\vartheta\ /\ -$",
        "y-axis": r"$X_{EO}\ /\ -$"
    }
])

show_table(dimensionless_base_plots, "Dimensionless base-case plots")

dimensionless_parameter_plots = pd.DataFrame([
    {
        "Plot": r"$X_{EO}(\zeta)$ for varying $Da_I$",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$X_{EO}\ /\ -$"
    },
    {
        "Plot": r"$\vartheta(\zeta)$ for varying $Da_I$",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$\vartheta\ /\ -$"
    },
    {
        "Plot": r"$X_{EO}(\zeta)$ for varying $St$",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$X_{EO}\ /\ -$"
    },
    {
        "Plot": r"$\vartheta(\zeta)$ for varying $St$",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$\vartheta\ /\ -$"
    },
    {
        "Plot": r"$\vartheta(\zeta)$ for varying $\beta$",
        "x-axis": r"$\zeta\ /\ -$",
        "y-axis": r"$\vartheta\ /\ -$"
    },
    {
        "Plot": r"$X_{EO}(\vartheta)$ for varying cooling intensity",
        "x-axis": r"$\vartheta\ /\ -$",
        "y-axis": r"$X_{EO}\ /\ -$"
    },
    {
        "Plot": r"$X_{EO,out}$ versus $Da_I$",
        "x-axis": r"$Da_I\ /\ -$",
        "y-axis": r"$X_{EO,out}\ /\ -$"
    },
    {
        "Plot": r"$\vartheta_{max}$ versus $St$",
        "x-axis": r"$St\ /\ -$",
        "y-axis": r"$\vartheta_{max}\ /\ -$"
    }
])

show_table(dimensionless_parameter_plots, "Dimensionless parameter-study plots")
```
